```python
import time
user_name= 'Anand'
password= 'hello@123'

max_attempts=3
attempts=0

print('enter your credentials')

while attempts < max_attempts:
  UserId=input('Enter your username: ')
  passcode=input('Enter your password: ')

  if UserId == user_name and passcode == password:
    print('login successful')
    break

  elif user_name!= UserId or passcode != password:
   attempts +=1
   remaining = max_attempts- attempts
   if remaining > 0:
     print(f'invalid username and password. Please try again. You have {remaining} attempts remaining')
     time.sleep(1)
   else:
     print("You're account has been locked. Please reset your password.")
  else:
      print('invalid username and password')
      break
print('Would you like to reset your password')
ans= input("Enter 'Y' for yes or 'N' for No")


```
